```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression #202431192_Glenn
```

```python
df = pd.DataFrame([[153,63], [174,81], [136,56],
                   [186,91], [128,47], [136,57],
                   [179,76], [163,72], [131,48]])
df.columns = ["x", "y"]
```

```python
df.corr()
```

```python
# menabagi data 
x_train = df['x'].values[:,np.newaxis]
y_train = df['y'].values

x_train
```

```python
y_train 
```

```python
# model training
lm = LinearRegression()
lm.fit(x_train, y_train)
```

```python
#membuat prediksi
x_test = [[170], [160], [180], [171], [150]]
p = lm.predict(x_test)
p
```

```python
# koefisien regresi & intersep
print("coefficient :", lm.coef_)
print("intercept :", lm.intercept_)
```

```python
#visualisasi model
pb = lm.predict(x_train)
dfc = pd.DataFrame({"x" : df["x"], "y" : pb})
plt.scatter(df['x'], df['y'])
plt.plot(dfc['x'], dfc['y'], color="magenta", linewidth=1)
plt.xlabel('Nilai X')
plt.ylabel('Nilai Y')
```

```python
df = pd.read_csv('botak.csv')
df
```

```python
df.info()
```

```python
import seaborn as sns
num_df = df.select_dtypes(include=['int64', 'float64'])

plt.figure(figsize=(10,6))
sns.heatmap(num_df.corr(), annot=True, cmap="coolwarm")
plt.show()
```

```python
db_clean = df.dropna(subset=["umur", "gaji", "tinggi", "berat", "botak_prob"])

x = db_clean[["umur", "gaji", "tinggi", "berat"]]
y = db_clean["botak_prob"]
```

```python
#memisah data menjadi data latih dan data test
from sklearn.model_selection import train_test_split as tts

x_train, x_test, y_train, y_test = tts(x, y, test_size=0.2, random_state=0)
```

```python
model = LinearRegression()

model.fit(x_train, y_train)
```

```python
# prediksi
y_pred = model.predict(x_test)
```

```python
#evaluasi
from sklearn.metrics import mean_squared_error, r2_score

print("coefficient :", model.coef_)
print("intercept :", model.intercept_)
print("MSE :", mean_squared_error(y_test, y_pred))
print("R2 Score :", r2_score(y_test, y_pred))
```

```python
#visualisasi model

plt.figure(figsize=(8,6))
plt.scatter(y_test, y_pred, alpha = 0.5)
plt.xlabel("Nilai asli y_test")
plt.ylabel("Nilai prediksi y_test")
plt.title("Nilai asli Vs Nilai prediksi")
plt.plot([0,1],[0,1],'r--')
plt.show()
```
